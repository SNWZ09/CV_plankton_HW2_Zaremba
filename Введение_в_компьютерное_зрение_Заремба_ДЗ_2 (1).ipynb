```python
#создаю пустой список и с помощью цикла помещаю в него все шесть изображений
pictures = []

for i in range(1, 7):
  pic_ex_name = f'/content/PLANKTON/Bosmina longirostris {i}.jpg'
  pic_ex = cv2.imread(pic_ex_name)
  pictures.append(pic_ex)

print(len(pictures))
```

```python
#вывожу полученный список, чтобы проверить, что все изображения туда попали
plt.figure(figsize=(18, 10))

for i, picture in enumerate(pictures):
    picture_rgb = cv2.cvtColor(picture, cv2.COLOR_BGR2RGB)

    plt.subplot(2, 3, i + 1)
    plt.imshow(picture_rgb)
    plt.title(f'Picture {i + 1}')
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
#опять создаю пустой список, но теперь для изображений в оттенках серого,
#так как выбранный мной метод SIFT анализирует яркость и структуру деталей
pictures_gray = []

plt.figure(figsize=(18, 10))

for i, picture in enumerate(pictures):
    picture_gray = cv2.cvtColor(picture, cv2.COLOR_BGR2GRAY)
    pictures_gray.append(picture_gray)

    plt.subplot(2, 3, i + 1)
    plt.imshow(picture_gray, cmap='gray')
    plt.title(f'Picture {i + 1}')
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
#создаю детектор
detector_sift = cv2.SIFT_create()
```

```python
#вычисляю ключевые точки и их дескрипторы
points_sift, descriptors_sift = detector_sift.detectAndCompute(pictures_gray[0], None)

print(len(points_sift))
print(descriptors_sift.shape)
```

```python
#рисую найденные ключевые точки на первом изображении для дальнейшего анализа
picture_with_points = cv2.drawKeypoints(
    pictures_gray[0],
    points_sift,
    None,
    flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
)
```

```python
#меняю порядок цветов для Матплотлиб (чтобы потом визуализировать результаты)
picture_with_points_rgb = cv2.cvtColor(picture_with_points, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(20, 9))

plt.subplot(1, 2, 1)
plt.imshow(pictures_gray[0], cmap='gray')
plt.title(f'Picture 1')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(picture_with_points_rgb)
plt.title(f'Picture 1 with keypoints (SIFT)')
plt.axis('off')

#вывожу исходное изображение в оттенках серого (слева) и изображение с ключевыми точками (справа)
plt.show()
```

```python
#перебор параметра contrastThreshold
#код тот же самый, что и был, но совмещен в одну ячейку
contrast_threshold_list = [0.02, 0.04, 0.06, 0.08]

for threshold in contrast_threshold_list:

  detector_sift_selection = cv2.SIFT_create(contrastThreshold=threshold)

  points_sift_selection, descriptors_sift_selection = detector_sift_selection.detectAndCompute(pictures_gray[0], None)

  print(len(points_sift_selection))
  print(descriptors_sift_selection.shape)

  picture_with_points_selection = cv2.drawKeypoints(
      pictures_gray[0],
      points_sift_selection,
      None,
      flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
  )

  picture_with_points_rgb_selection = cv2.cvtColor(picture_with_points_selection, cv2.COLOR_BGR2RGB)

  plt.figure(figsize=(20, 9))
  plt.suptitle(f'Contrast threshold = {threshold}', fontsize=32, y=0.9)
  plt.axis('off')

  plt.subplot(1, 2, 1)
  plt.imshow(pictures_gray[0], cmap='gray')
  plt.axis('off')

  plt.subplot(1, 2, 2)
  plt.imshow(picture_with_points_rgb_selection)
  plt.axis('off')


  plt.show()
```

```python
#перебор параметра edgeThreshold
#код тот же самый, что и был, но совмещен в одну ячейку
edge_threshold_list = [5, 10, 20]

for edge_threshold in edge_threshold_list:

  detector_sift_selection = cv2.SIFT_create(edgeThreshold=edge_threshold)

  points_sift_selection, descriptors_sift_selection = detector_sift_selection.detectAndCompute(pictures_gray[0], None)

  print(len(points_sift_selection))
  print(descriptors_sift_selection.shape)

  picture_with_points_selection = cv2.drawKeypoints(
      pictures_gray[0],
      points_sift_selection,
      None,
      flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
  )

  picture_with_points_rgb_selection = cv2.cvtColor(picture_with_points_selection, cv2.COLOR_BGR2RGB)

  plt.figure(figsize=(20, 9))
  plt.suptitle(f'Edge threshold = {edge_threshold}', fontsize=32, y=0.9)

  plt.subplot(1, 2, 1)
  plt.imshow(pictures_gray[0], cmap='gray')
  plt.axis('off')

  plt.subplot(1, 2, 2)
  plt.imshow(picture_with_points_rgb_selection)
  plt.axis('off')


  plt.show()
```

```python
#перебор параметра sigma
#код тот же самый, что и был, но совмещен в одну ячейку
sigma_selection_list = [0.8, 1.2, 1.6, 2.0]

for sigma_selection in sigma_selection_list:

  detector_sift_selection = cv2.SIFT_create(sigma=sigma_selection, contrastThreshold=0.04, edgeThreshold=10)

  points_sift_selection, descriptors_sift_selection = detector_sift_selection.detectAndCompute(pictures_gray[0], None)

  print(len(points_sift_selection))
  print(descriptors_sift_selection.shape)

  picture_with_points_selection = cv2.drawKeypoints(
      pictures_gray[0],
      points_sift_selection,
      None,
      flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
  )

  picture_with_points_rgb_selection = cv2.cvtColor(picture_with_points_selection, cv2.COLOR_BGR2RGB)

  plt.figure(figsize=(20, 9))
  plt.suptitle(f'Sigma value = {sigma_selection}', fontsize=32, y=0.9)

  plt.subplot(1, 2, 1)
  plt.imshow(pictures_gray[0], cmap='gray')
  plt.axis('off')

  plt.subplot(1, 2, 2)
  plt.imshow(picture_with_points_rgb_selection)
  plt.axis('off')


  plt.show()
```

```python
#перебор параметра nOctaveLayers
#код тот же самый, что и был, но совмещен в одну ячейку
layers_selection_list = [2, 3, 4, 5]

for octave_layer in layers_selection_list:

  detector_sift_selection = cv2.SIFT_create(nOctaveLayers=octave_layer, sigma=1.6, contrastThreshold=0.04, edgeThreshold=10)

  points_sift_selection, descriptors_sift_selection = detector_sift_selection.detectAndCompute(pictures_gray[0], None)

  print(len(points_sift_selection))
  print(descriptors_sift_selection.shape)

  picture_with_points_selection = cv2.drawKeypoints(
      pictures_gray[0],
      points_sift_selection,
      None,
      flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
  )

  picture_with_points_rgb_selection = cv2.cvtColor(picture_with_points_selection, cv2.COLOR_BGR2RGB)

  plt.figure(figsize=(20, 9))
  plt.suptitle(f'Octave layers= {octave_layer}', fontsize=32, y=0.9)

  plt.subplot(1, 2, 1)
  plt.imshow(pictures_gray[0], cmap='gray')
  plt.axis('off')

  plt.subplot(1, 2, 2)
  plt.imshow(picture_with_points_rgb_selection)
  plt.axis('off')


  plt.show()
```

```python
#нахожу точки и вычисляю дескрипторы для каждого из шести изображений
all_points = []
all_descriptors =[]

all_pic_iteration_detector_sift = cv2.SIFT_create(nOctaveLayers=3, sigma=1.6, contrastThreshold=0.04, edgeThreshold=10)


for picture in pictures_gray:

  points_sift, descriptors_sift = all_pic_iteration_detector_sift.detectAndCompute(picture, None)

  all_points.append(points_sift)
  all_descriptors.append(descriptors_sift)

  print('Points value', len(points_sift))
  print('Descriptor shape', descriptors_sift.shape)
```

```python
#ищу похожие дескрипторы между первым и вторым изображением
descriptor_match = cv2.BFMatcher(cv2.NORM_L2)
best_match = descriptor_match.knnMatch(all_descriptors[0], all_descriptors[1], k=2)
```

```python
print(len(best_match))
```

```python
#провожу тест Лоу, 'чтобы из всех
#кандидатов оставить только геометрически устойчивые и информативные точки'
#(с платформы, урок 'Сравнение, сопоставление и нормализация изображений')
lowe_test_matches = []

for i, j in best_match:
  if i.distance < 0.7 * j.distance:
    lowe_test_matches.append(i)

print(len(lowe_test_matches))
```

```python
#создаю сравнение двух изображений с линиями, соединяющими похожие точки
comparison = cv2.drawMatches(
    pictures[0], all_points[0], pictures[1], all_points[1],
    lowe_test_matches, None,
    matchesThickness=8,
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
)
```

```python
#вывожу это сравнение
comparison_picture = cv2.cvtColor(comparison, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(20, 9))
plt.suptitle('SIFT Keypoints comparison (1-2 pictures)', fontsize=22, y=0.95)
plt.imshow(comparison_picture)
plt.axis('off')

plt.show()
```

```python
#сравниваю дескрипторы для пар изображений 1-2, 1-3, 1-4, 1-5, 1-6,
#провожу тесты лоу, сортирую топ 30 лучших по расстаянию пар и визуализирую это
descriptor_match = cv2.BFMatcher(cv2.NORM_L2)

all_pair_matches = []

for index in range(1, 6):

    best_match = descriptor_match.knnMatch(all_descriptors[0], all_descriptors[index], k=2)

    lowe_test_matches = []

    for i, j in best_match:
        if i.distance < 0.7 * j.distance:
            lowe_test_matches.append(i)

    all_pair_matches.append(lowe_test_matches)

    best_30_matches = sorted(lowe_test_matches, key=lambda match: match.distance)[:30]

    print(f'Picture 1 and {index + 1}: '
        f'after Lowe-Test — {len(lowe_test_matches)}, ')

    comparison = cv2.drawMatches(
        pictures[0], all_points[0],
        pictures[index], all_points[index],
        best_30_matches, None,
        matchesThickness=6,
        flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS)

    comparison_picture = cv2.cvtColor(comparison, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(20, 10))
    plt.imshow(comparison_picture)
    plt.title(f'Picture 1 and {index + 1}: top-30 best matches', fontsize=22)
    plt.axis('off')
    plt.show()
```

```python
#тут ищу количество точек у первого изображения, у которых в каждом другом изображении есть пара (после теста Лоу)
common_points = set()

for match in all_pair_matches[0]:
    common_points.add(match.queryIdx)

for matches in all_pair_matches[1:]:
    current_points = set()

    for match in matches:
        current_points.add(match.queryIdx)

    common_points = common_points.intersection(current_points)

print('Points in common:', len(common_points))
```
